**Q41.** Take two numbers from the user (as integers) and print their sum, difference, product and division.

````python
a = int(input('Enter first number: '))
b = int(input('Enter second number: '))
print('Sum:', a + b)
print('Difference:', a - b)
print('Product:', a * b)
print('Division:', a / b)
````

**Q42.** Take the radius of a circle from the user (as a float) and print its area. Use `area = 3.14 * r**2`.

````python
r = float(input('Enter radius: '))
area = 3.14 * r**2
print('Area of circle:', area)
````

**Q43.** Take a total number of seconds from the user and print it as minutes and remaining seconds. (Hint: use `//` and `%`.)

````python
total = int(input('Enter seconds: '))
minutes = total // 60
seconds = total % 60
print(minutes, 'minutes', seconds, 'seconds')
````

**Q44.** Take a number from the user and print `True` if it is even, otherwise `False`. (Hint: `%` and `==`.)

````python
num = int(input('Enter a number: '))
print(num % 2 == 0)
````

**Q45.** Compound interest: take principal `P`, rate `r` (in %) and years `t` from the user and print the final amount `P*(1+r/100)**t`.

````python
P = float(input('Principal: '))
r = float(input('Rate (%): '))
t = int(input('Years: '))
amount = P * (1 + r/100)**t
print('Final amount:', amount)
````

**Q46.** Swap the values of two variables `x = 'Gen AI'` and `y = 'Agentic AI'` using a third variable, and print them.

````python
x = 'Gen AI'
y = 'Agentic AI'
temp = x
x = y
y = temp
print(x, y)
````

**Q47.** Take the user's name and age, then print the name and the age they will be after 5 years.

````python
name = input('Enter your name: ')
age = int(input('Enter your age: '))
print(name, 'will be', age + 5, 'after 5 years')
````

**Q48.** Take a number from the user and print its last digit. (Hint: `% 10`.)

**Q49.** Take a number from the user and print its square, square root and cube root using only the `**` operator.

````python
n = float(input('Enter a number: '))
print('Square:', n**2)
print('Square root:', n**0.5)
print('Cube root:', n**(1/3))
````

**Q50.** Take your marks and your friend's marks from the user. Print the results of `>`, `<`, `==` and `!=` between them, and the type of one result.

````python
my_marks = int(input('Your marks: '))
friend_marks = int(input("Friend's marks: "))
print('Greater?', my_marks > friend_marks)
print('Smaller?', my_marks < friend_marks)
print('Equal?', my_marks == friend_marks)
print('Not equal?', my_marks != friend_marks)
print(type(my_marks > friend_marks))
````
